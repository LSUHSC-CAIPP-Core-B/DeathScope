# DeathScope YOLO Training
Interactive example for training and evaluating the DeathScope detector with Ultralytics.

```python
from ultralytics import YOLO
```

## Load Base Model

```python
model = YOLO("yolov8l.pt")
```

## Train Model

```python
model.train(data="cell_dataset.yaml", 
            epochs=100,
            batch=8, 
            imgsz=128,
            project="runs/train",
            name="deathscope_yolov8l")
```

## Metrics

```python
model = YOLO('runs/train/deathscope_yolov8l/weights/best.pt')  # load a trained model

# Validate the model
metrics = model.val()  # no arguments needed, dataset and settings remembered
print(metrics.box.map)    # map50-95
print(metrics.box.map50)  # map50
print(metrics.box.map75)  # map75
print(metrics.box.maps)   # a list contains map50-95 of each category
```

## Inference

```python
# Load the trained model
model = YOLO('runs/train/deathscope_yolov8l/weights/best.pt')

# Define path to directory containing images and videos for inference
source = 'path/to/dir'

# Run inference on the source
results = model(source, stream=True)  # generator of Results objects

# Process results list
for result in results:
    boxes = result.boxes  # Boxes object for bbox outputs
```
